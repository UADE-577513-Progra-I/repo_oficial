```python
ratings = [5, 3, 4, 1, 5, 2, 4, 3, 5, 1, 4, 2]
ratings
```

```python
# listas por comprension
[rating*2 for rating in ratings]
```

```python
# map + lambda
print(ratings)
resultado = map( lambda rating : rating , ratings)
print(f"resultado {resultado}")
print(f"id_resultado {id(resultado)}")
print(next(resultado))  # 5
print(next(resultado))  # 3
print(next(resultado))  # 4
```

```python
def multiplicar(x):
    return x*2

multiplicar = lambda x : x*2

resultado = list(map( lambda r : r*2 , ratings))
print(resultado)
```

```python
# lista por comprension
# filtrar los elementos mayores a 4
[ r  for r in ratings if r>=4 ]
```

```python
# filter + lambda
# filtrar los elementos mayores a 4

#list(filter(  funcion , iterable))
#list(filter(  lambda  parametro : condicion , iterable))
list(filter(  lambda  r : r>=4 , ratings))
```
